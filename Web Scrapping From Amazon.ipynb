```python
from playwright.sync_api import sync_playwright
import pandas as pd
import time
import nest_asyncio
nest_asyncio.apply()
```

```python
brand_name = []
price = []
rating_review = []
star_rating = []
discount = []
other_info = []
```

```python
with sync_playwright() as p:
    browser = p.chromium.launch(headless=False)
    page = browser.new_page()

    for i in range(1, 11):
        print(f'loading {i}')
        page.goto('https://www.flipkart.com/search?q=mobile+5g&sid=tyy%2C4io&as=on&as-show=on&otracker=AS_QueryStore_OrganicAutoSuggest_1_6_na_na_na&otracker1=AS_QueryStore_OrganicAutoSuggest_1_6_na_na_na&as-pos=1&as-type=RECENT&suggestionId=mobile+5g%7CMobiles&requestId=2767a307-1eb1-4b18-9a87-588f2d22f48c&as-searchtext=mobile%205g&page={}'.format(i))

        time.sleep(2)

        products = page.locator('.ZFwe0M.row')
        count = products.count()

        for j in range(count):
            p_locator = products.nth(j)
            
            # brand name 
            try:
                brand_name.append(p_locator.locator('.RG5S1k').inner_text())
            except:
                brand_name.append(None)
                
            # Price 
            try:
                price.append(p_locator.locator('.hZ3P6w.DeU9vF').inner_text())
            except:
                price.append(None)
                
            # Rating & Review
            try:
                rating_review.append(p_locator.locator('.Wphh3N').inner_text())
            except:
                rating_review.append(None)
                
            # Discount
            try:
                discount.append(p_locator.locator('.HQe8jr').inner_text())
            except:
                discount.append(None)
                
            # Other Info (RAM/ROM)
            try:
                other_info.append(p_locator.locator('.HwRTzP').inner_text())
            except:
                other_info.append(None)

    browser.close()

```
