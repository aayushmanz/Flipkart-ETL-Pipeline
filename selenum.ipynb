```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
import time
```

```python
brand_name = []
price = []
rating_review = []
star_rating = []
discount = []
other_info = []
```

```python
driver = webdriver.Chrome()

for i in range(1, 25):
 print(f"Scrapping page {i}")  
  
 driver.get('https://www.flipkart.com/search?q=mobile+5g&sid=tyy%2C4io&as=on&as-show=on&otracker=AS_QueryStore_OrganicAutoSuggest_1_6_na_na_na&otracker1=AS_QueryStore_OrganicAutoSuggest_1_6_na_na_na&as-pos=1&as-type=RECENT&suggestionId=mobile+5g%7CMobiles&requestId=2767a307-1eb1-4b18-9a87-588f2d22f48c&as-searchtext=mobile%205g&page={}'.format(i))
 time.sleep(3)
 products = driver.find_elements(By.CSS_SELECTOR, '.ZFwe0M.row')
for p in products:
    try:
     brand_name.append(p.find_element(By.CLASS_NAME, 'RG5Slk').text)
    except:
        brand_name.append(None)

    try:
     price.append(p.find_element(By.CLASS_NAME, 'QiMO5r').text)
    except:
     price.append(None)   

    try:
     rating_review.append(p.find_element(By.CLASS_NAME, 'PvbNMB').text)
    except:
       rating_review.append(None)

    try:
     star_rating.append(p.find_element(By.CLASS_NAME, 'MKiFS6').text)
    except:
        star_rating.append(None) 

    try:
     discount.append(p.find_element(By.CLASS_NAME, 'HQe8jr').text)
    except:
        discount.append(None)  

    try:
     other_info.append(p.find_element(By.CLASS_NAME, 'HwRTzP').text)
    except:
        other_info.append(None)            
```

```python
# unfortunately price is showing NaN and None so i m skipping it !
```

```python
data = {
    "brand name" : brand_name,
    "rating & review" : rating_review,
    "star rating" : star_rating,
    "discount" : discount,
    "other info" : other_info
}


df = pd.DataFrame(data)

df
```
