```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('flipkart_mobile.csv')
df.head(10)
```

```python
df = df.drop(columns=['Unnamed: 0'])
```

```python
df.insert(0, 'model name' ,df['brand name'].str.split('(').str[0])
```

```python
df
```

```python
temp = df['brand name'].str.split('(').str[1]
color = temp.str.split(',').str[0]


temp = temp.str.split(',').str[1]
storage = temp.str.replace(')', '')


df.insert(1, 'model color', color)
df.insert(2, 'storage', storage)

```

```python
df = df.drop(columns='brand name')
```

```python
df
```

```python
rating = df['rating & review'].str.split('&').str[0]
rating = rating.str.replace('Ratings', '').str.strip()

review = df['rating & review'].str.split('&').str[1]
review = review.str.replace('Reviews', '').str.strip()

df.insert(3, 'review', review)
df.insert(4, 'rating', rating)
```

```python
df.drop(columns='rating & review', inplace=True)
```

```python
df.replace('NaN', np.nan, inplace=True)
```

```python
# because if no one give review and rating the value will show 0 thats why i removed the null 

df[['review', 'rating']] = df[['review', 'rating']].replace(['NaN', np.nan], '0')
```

```python
df
```

```python
df['star rating'] = df['star rating'].replace(np.nan, 0.0)
```

```python
df
```

```python
df['discount']= df['discount'].replace(np.nan, 'No Discount')
```
